# Missing value

```python
data = {
    'age':[25,np.nan,34,27],
    'score':[56,78,np.nan,np.nan],
    'City':['jetpur',np.nan,np.nan,'rajkot']
}
dataf = pd.DataFrame(data)
dataf
```

```python
dataf.isnull()
# dataf.isnull().sum()
```

```python
dataf.notnull()
# dataf.notnull().sum()
```

```python
df_row = dataf.dropna()
df_row
```

```python
df_row = dataf.dropna(axis=1)
df_row
```

```python
df_row = dataf.dropna(subset=['age','City'])
df_row
```

## Advanced Machine Learning Imputation

```python
# data = {
#     'age':[25,np.nan,34,27],
#     'score':[56,78,np.nan,np.nan],
#     'City':['jetpur',np.nan,np.nan,'rajkot']
# }
# dataf = pd.DataFrame(data)
# dataf

from sklearn.impute import SimpleImputer, KNNImputer

mean_inputer = SimpleImputer(strategy='mean')
dataf[['age']] = mean_inputer.fit_transform(dataf[['age']])

knn_imputer = KNNImputer(n_neighbors=2)
dataf[['score']] = knn_imputer.fit_transform(dataf[['score']])
```

```python
dataf[['age','score']]
```

```python
# data = {
#     'age':[25,np.nan,34,27],
#     'score':[56,78,np.nan,np.nan],
#     'City':['jetpur',np.nan,np.nan,'rajkot']
# }
# dataf = pd.DataFrame(data)
# dataf

from sklearn.impute import SimpleImputer, KNNImputer

mean_inputer = SimpleImputer(strategy='median')
dataf[['age']] = mean_inputer.fit_transform(dataf[['age']])

knn_imputer = KNNImputer(n_neighbors=2)
dataf[['score']] = knn_imputer.fit_transform(dataf[['score']])
```

## Removing Dupllicates

```python
data = {
    'Name':['prince','mayur','badi','smit','smit','badi','badi'],
    'Age':[21,20,34,19,21,20,20],
    'City':['jetpur','botad','vadhvan','jetpur','botad','jetpur','jetpur']
}

df = pd.DataFrame(data)
```

```python
df
# print("/n" + "=" * 40 + "/n")
```

```python
dfr = df.drop_duplicates()
dfr
```

```python
# dfr = df.drop_duplicates(subset=['City'])
dfr = df.drop_duplicates(subset=['Age','City'])
dfr
```

```python
# dfr = df.drop_duplicates(subset=['City'])
dfr = df.drop_duplicates(subset=['Name'], keep = 'last')
dfr
```

```python
# dfr = df.drop_duplicates(subset=['City','Age'])
dfr = df.drop_duplicates(subset=['Name'], keep = False)
dfr
```

```python
# dfr = df.drop_duplicates(subset=['City','Age'])
dfr = df.drop_duplicates(inplace = True, ignore_index = True)
dfr
```
