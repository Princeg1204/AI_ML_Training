```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
```

```python
df.head()
```

## Univariate Analysis

```python
print(df['total_bill'].describe())

plt.figure(figsize = (8,4))
sns.histplot(df['total_bill'], kde = True, color = 'y')
plt.title('Univariate : Distribution od total bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
print(df['day'].value_counts())

plt.figure(figsize = (8,4))
sns.countplot(data = df, x = 'day', palette = 'Set1')
plt.title('Univariate : Order Frequency by Day')
plt.xlabel('Day of week')
plt.ylabel('Count')
plt.show()
```

## Bivariate Analysis

```python
plt.figure(figsize = (7,5))
sns.regplot(data = df, x = 'total_bill', y = 'tip', scatter_kws = {'alpha':0.7})
plt.title('Bivariate : Total Bill vs. Tip Amount')
plt.xlabel('Total bill ($)')
plt.ylabel('Tip ($)')
plt.show()

correlation = df['total_bill'].corr(df['tip'])
print(f"Correlation between Total Bill and Tip : {correlation: .2f}")
```

## Categorical vs. Numerical

```python
plt.figure(figsize = (8,5))
sns.boxplot(data = df, x='day', y='total_bill', palette='Set1')
plt.title('Bivariate : Total Bill by Day of Week')
plt.xlabel('Day')
plt.ylabel('Total Bill ($)')
plt.show()
```

## Multivariate Analysis

```python
plt.figure(figsize = (8,5))
sns.scatterplot(
    data =df,
    x='total_bill',
    y='tip',
    hue='smoker',
    style='smoker',
    s=70,
)
plt.title('Multivariate : Bill vs. Tip by Smoker Status')
plt.ylabel('Tip ($)')
plt.xlabel('Total Bill ($)')
plt.show()
```

```python
numeric_df = df.select_dtypes(include=['float64','int64'])
corr_martix = numeric_df.corr()

plt.figure(figsize = (6,5))
sns.heatmap(corr_martix, annot = True, cmap ='coolwarm',vmin = -1, vmax = 1)
plt.title('Multivariate: Correlation Matrix Heatmap')
plt.show()
```
