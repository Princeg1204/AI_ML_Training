```python
x = df['Rating']
y = df['Reviews']
```

```python
plt.scatter(x,y)
```

```python
plt.scatter(x,y)
plt.title('Google playstore scatter plot')
plt.xlabel('Rating')
plt.ylabel('Reviews')
# plt.show()
```

```python
plt.scatter(x,y, c = 'purple', marker = '*')
plt.title('Google playstore scatter plot')
plt.xlabel('Rating')
plt.ylabel('Reviews')
# plt.show()
```

```python
plt.figure(figsize = (16,9))
plt.scatter(x,y, c = 'r', marker = '*')
plt.title('Google playstore scatter plot')
plt.xlabel('Rating')
plt.ylabel('Reviews')
# plt.show()
```

```python
plt.figure(figsize = (16,9))
plt.scatter(x,y, c = 'r', marker = '*', alpha = 0.1)
plt.title('Google playstore scatter plot')
plt.xlabel('Rating')
plt.ylabel('Reviews')
# plt.show()
```

```python
plt.figure(figsize = (16,9))
plt.scatter(x,y, c = 'r', marker = '*', s = 200, alpha = None,
           linewidth = 2, edgecolors = 'k')
plt.title('Google playstore scatter plot')
plt.xlabel('Rating')
plt.ylabel('Reviews')
# plt.show()
```

```python
plt.figure(figsize = (16,9))

plt.scatter(x,y, c = 'r', marker = '*', s = 150, alpha = 0.5,
           linewidth = 1, edgecolors = 'k')
plt.scatter(x,df['Installs'], c = 'white', marker = 'o', s = 150, alpha = 0.5,
           linewidth = 1, edgecolors = 'b')

plt.title('Google playstore scatter plot')
plt.xlabel('Rating')
plt.ylabel('Reviews')
# plt.show()
```
