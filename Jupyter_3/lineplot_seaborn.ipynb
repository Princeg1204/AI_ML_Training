```python
tips_df = sns.load_dataset('tips')
tips_df
```

```python
 sns.lineplot(x='size', y='tip',data=tips_df)
```

```python
 sns.lineplot(x='size', y='total_bill',hue = 'time',style = 'time',data=tips_df)
```
