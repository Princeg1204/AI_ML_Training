# **Funtions**

```python
def calculate_margin(rev,exp):
  profit = rev - exp
  margin = profit * 100 / rev
  return margin
```

```python
calculate_margin(50, 10)
```

```python
def calculate_profit_and_margin(rev,exp):
  profit = rev - exp
  margin = profit * 100 / rev
  return profit,margin
```

```python
response = calculate_profit_and_margin(50, 10)
response
```

```python
from datetime import date

class Human:
    def __init__(self,name,birth_year):
        self.name = name
        self.birth_year = birth_year

    def calculate_birthyear(self):
        current_year = date.today().year
        return current_year - self.birth_year

    def introduce(self):
        print(f"Hi, my name is {self.name()}. and I am {self.birth_year()} years old.") 

emp_1 = Human('Prince',21)
emp_1.calculate_birthyear()
   
```
