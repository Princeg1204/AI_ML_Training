```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    'Gender' : ['M','F','F','M','F','M']
})
df
```

```python
encoder = LabelEncoder()
df['Gender'] = encoder.fit_transform(df['Gender'])
print(df)
```

```python
df1 = pd.DataFrame({
    'City' : ['Pune','Delhi','Mumbai','Banglore']
})
df1
```

```python
encode = pd.get_dummies(df1, columns = ['City'], dtype = int)
print(encode)
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output = False)
result = encoder.fit_transform(df1[['City']])
print(result)
```

```python
from sklearn.preprocessing import OrdinalEncoder
df2 = pd.DataFrame({
    'Education':['Graduate','PhD','High School','Post Graduate']
})
df2
```

```python
encoder = OrdinalEncoder(categories = [[
    'Graduate','PhD','High School','Post Graduate'
]])
df2['Education'] = encoder.fit_transform(df2[['Education']])
print(df2)
```

```python
df3 = pd.DataFrame({
    "City":[
        "Pune",
        "Delhi",
        "Pune",
        "Mumbai",
        "Pune"
    ]
})

df3
```

```python
frequency = df3["City"].value_counts()

df3["City"] = df3["City"].map(frequency)

print(df3)
```

```python
df4 = pd.DataFrame({
    "City":["Pune","Delhi","Pune","Delhi"],
    "Price":[60,80,55,90]
})

df4
```

```python
mean = df4.groupby("City")["Price"].mean()
mean
```

```python
df4['City'] = df4['City'].map(mean)
df4
```
