```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Color':['red','green','blue']
})
pd.get_dummies(df)
```

```python
pd.get_dummies(df, dtype = int)
```

```python
pd.get_dummies(df, dtype = int)
```

```python
pd.get_dummies(df, prefix = 'clr', dtype = int)
```

```python
df1 = pd.DataFrame({
    'color': ['blue','green'],
    'size': ['s' ,'m']
})

pd.get_dummies(df1)
```

```python
pd.get_dummies(df1, prefix = ['c' , 's'], dtype = int)
```

```python
pd.get_dummies(df1, prefix = 'color', prefix_sep = '-', dtype = int)
```

```python
df2 = pd.DataFrame({
    'Color': ['Blue', 'Green', None, 'Pink']
})
pd.get_dummies(df2, dtype = int)
```

```python
df3 = pd.DataFrame({
    'Color': ['black', 'white', 'pink'],
    'Age': [21,25,26],
    'Gender':['M', 'F' , 'F']
})
pd.get_dummies(df3, dtype = int)
```

```python
pd.get_dummies(df3, columns = ['Color'], dtype = int)
```

```python
pd.get_dummies(df3, columns = ['Color','Gender'], dtype = int)
```

```python
pd.get_dummies(df3, sparse = True, dtype = int)
```

```python
df4 = pd.DataFrame({
    'Gender':['Male','Female','Male']
})
```

```python
pd.get_dummies(df4, dtype = int)
```

```python
pd.get_dummies(df4, drop_first = True, dtype = int)
```
