```python
classes = ['python', 'AI', 'ML', 'DS']
class1_student = [29,67,45,32,69]
explode = [0,0,0.2,0,0]
colors = ['c','b','r','y','g']
plt.pie(class1_student, explode = explode, colors = colors, autopct = '% 0.1f %%', shadow = True)
```

```python
classes = ['OS','python', 'AI', 'ML', 'DS']
class1_student = [29,67,45,32,69]
explode = [0,0,0.2,0,0]
colors = ['c','b','r','y','g']
plt.pie(class1_student, explode = explode, labels = classes, colors = colors, 
        autopct = '% 0.1f %%', shadow = True, radius = 1.5, startangle = 280)
```

```python
classes = ['OS','python', 'AI', 'ML', 'DS']
class2_student = [34,56,74,35,31]
explode = [0,0,0.2,0,0.1]
colors = ['c','b','r','y','g']
textprops = {'fontsize': 60}
plt.pie(class2_student, explode = explode, labels = classes, colors = colors, 
        autopct = '% 0.1f %%', shadow = True, radius = 1.4, startangle = 280)
```

```python
plt.figure(figsize = (16,9))
wedgeprops = {'linewidth' : 3, 'width': 1, 'edgecolor':'k'}
plt.pie(class1_student, explode = explode, labels = classes, colors = colors, 
        autopct = '% 0.2f %%', shadow = True, radius = 1.1, startangle = 180,
       pctdistance = 0.8, labeldistance = 1.4, counterclock = True, wedgeprops = wedgeprops, textprops = textprops, center = (2,3), frame = True, rotatelabels = True)
plt.show()
```
