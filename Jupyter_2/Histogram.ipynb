```python
ml_students_age = np.random.randint(18,28,(100))
py_students_age = np.random.randint(14,28,(100)) #start age, end age, values

bins = [18,20,22,24,26,28]

# plt.hist(ml_students_age, bins, rwidth = 0.90, histtype = 'step')
plt.hist(ml_students_age, bins, rwidth = 0.90, histtype = 'barstacked')

plt.title('ML students age histogram')
plt.xlabel('Student age category')
plt.ylabel('No. student age')
plt.show()
```

```python
bins = [18,20,22,24,26,28]

# plt.hist(ml_students_age, bins, rwidth = 0.90, histtype = 'step')
plt.hist(ml_students_age, bins, rwidth = 0.90, color = 'm', width = 0.9, histtype = 'barstacked', align = 'left')

plt.title('ML students age histogram')
plt.xlabel('Student age category')
plt.ylabel('No. student age')
plt.show()
```

```python
bins = [18,20,22,24,26,28]

# plt.hist(ml_students_age, bins, rwidth = 0.90, histtype = 'step')
plt.hist(ml_students_age, bins, rwidth = 0.9, color = 'r', histtype = 'barstacked', orientation = 'horizontal')

plt.title('ML students age histogram')
plt.xlabel('Student age category')
plt.ylabel('No. student age')
plt.show()
```

```python
bins = [18,20,22,24,26,28]

# plt.hist(ml_students_age, bins, rwidth = 0.90, histtype = 'step')
plt.hist(ml_students_age, bins, rwidth = 0.90, color = 'c', histtype = 'barstacked', label = 'ML Students')

plt.title('ML students age histogram')
plt.xlabel('Student age category')
plt.ylabel('No. student age')
plt.legend()
plt.show()
```

```python
bins = [18,20,22,24,26,28]

# plt.hist(ml_students_age, bins, rwidth = 0.90, histtype = 'step')
plt.hist(ml_students_age, bins, rwidth = 0.90, color = 'y', histtype = 'barstacked', label = 'ML Students')
plt.hist(py_students_age, bins, rwidth = 0.90, color = 'b', histtype = 'barstacked', label = 'PY Students')

plt.title('ML students age histogram')
plt.xlabel('Student age category')
plt.ylabel('No. student age')
plt.legend()
plt.show()
```

```python
bins = [18,20,22,24,26,28]

# plt.hist(ml_students_age, bins, rwidth = 0.90, histtype = 'step')
plt.hist(ml_students_age, bins, rwidth = 0.90, color = 'y', histtype = 'barstacked', orientation = 'horizontal', label = 'ML Students')
plt.hist(py_students_age, bins, rwidth = 0.90, color = 'b', histtype = 'barstacked', orientation = 'vertical', label = 'PY Students')

plt.title('ML students age histogram')
plt.xlabel('Student age category')
plt.ylabel('No. student age')
plt.legend()
plt.show()
```

```python
bins = [18,20,22,24,26,28]

# plt.hist(ml_students_age, bins, rwidth = 0.90, histtype = 'step')
plt.hist([ml_students_age,py_students_age], bins, rwidth = 0.90, histtype = 'bar',color = ['r','b'], label = ['ML Students','PY Students'])

plt.title('ML students age histogram')
plt.xlabel('Student age category')
plt.ylabel('No. student age')
plt.legend()
plt.show()
```
