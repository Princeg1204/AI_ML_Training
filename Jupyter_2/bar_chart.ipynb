```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style
```

```python
classes = ['Python', 'R', 'AI', 'DSA', 'OOP']

class1_student = [29,67,45,32,69]
class2_student = [34,56,74,35,31]
class3_student = [65,89,90,21,40]

# plt.bar(classes, class1_student, label = 'Class1')
# plt.bar(classes, class2_student, label = 'Class2')
plt.barh(classes, class3_student, label = classes)
plt.legend()
plt.show()
```

```python
plt.bar(classes, class1_student, width = 0.2, align = 'edge', color = 'b', edgecolor = 'r', linewidth = 0.5, alpha = 0.5, linestyle = '--')
```

```python
plt.bar(classes, class2_student, width = 0.2, align = 'edge', color = 'c', edgecolor = 'r', 
        linewidth = 0.5, alpha = 0.5, linestyle = '--', label = 'class 2 student')
```

```python
style.use('ggplot')
plt.barh(classes, class3_student, align = 'edge', color = 'y', edgecolor = 'r', 
        linewidth = 0.5, alpha = 0.5, linestyle = '--', label = 'class 3 student')
plt.legend()
```

```python
plt.figure(figsize = (16,9))
plt.barh(classes, class3_student, align = 'edge', color = 'b', edgecolor = 'r', 
        linewidth = 0.5, alpha = 0.5, linestyle = '--', label = 'class 3 student')

plt.title('Bar chart of MU class', fontsize = 15)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No. of student', fontsize = 20)
plt.show()
```

```python
plt.figure(figsize = (16,9))

plt.bar(classes, class1_student, width = 0.4, color = 'b', label = 'class 1 student')

plt.bar(classes, class3_student, width = 0.4, color = 'c', label = 'class 3 student')

plt.title('Bar chart of MU class', fontsize = 15)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No. of student', fontsize = 20)
plt.show()
```

```python
plt.figure(figsize = (16,9))

classes_index = np.arange(len(classes))

width = 0.2

plt.bar(classes_index , class1_student, width, color = 'b')

plt.bar(classes_index + width, class2_student, width, color = 'c' )

plt.title('Bar chart of MU class', fontsize = 18)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No. of student', fontsize = 20)
# plt.show()
```

```python
plt.figure(figsize = (16,9))

classes_index = np.arange(len(classes))

width = 0.2

plt.bar(classes_index, class1_student, width, color = 'b')

plt.bar(classes_index + width, class3_student, width, color = 'c' )

plt.xticks(classes_index + width, classes, rotation = 20)

plt.title('Bar chart of MU class', fontsize = 18)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No. of student', fontsize = 20)
# plt.show()
```

```python
plt.figure(figsize = (16,9))

classes_index = np.arange(len(classes))

width = 0.2

plt.bar(classes_index, class1_student, width, color = 'b')

plt.bar(classes_index + width, class2_student, width, color = 'c' )

plt.bar(classes_index + width + width, class3_student, width, color = 'y' )

plt.xticks(classes_index + width, classes, rotation = 20)

plt.title('Bar chart of MU class', fontsize = 18)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No. of student', fontsize = 20)
# plt.show()
```
